```python
# Setting system path and project root
import os
import sys

PROJECT_ROOT_DIR = os.path.abspath('../..')
sys.path.append(PROJECT_ROOT_DIR) # bringing system path to project root

def get_fp(relative_path):
    return os.path.join(PROJECT_ROOT_DIR, relative_path)
```

```python
import pandas as pd

def read_jsonl_gz_lazy(path: str, chunksize: int = 10_000):
    json_iter = pd.read_json(
        path,
        compression='gzip',
        lines=True,
        chunksize=chunksize,
    )

    for chunk in json_iter:
        yield chunk
```

```python
jsonl_gz_filepath = get_fp('data_dir/fw_samples_new/samples/abk_Cyrl/00000.jsonl.gz')
for df_chunk in read_jsonl_gz_lazy(jsonl_gz_filepath, chunksize=5000):
        print(df_chunk.shape)
        print(df_chunk.head(2))
```
